```python
# pip install scikit-learn
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# load the data
df = pd.read_csv("../../../Datasets/adult.csv")


# let's quickly see the first 5 rows of data
df.head()
```

```python
df.info()
```

```python
# Let's check the shape of the data
df.shape
```

```python
#Checking if we have empty rows
df.isna().sum()
```

```python
# check duplicates 
df.duplicated().sum()
```

```python
#drop the duplicated rows
df = df.drop_duplicates()
```

```python
df.duplicated().sum()
```

```python
# checking how many columns have this value
(df == "?").sum()
```

```python
# Replacing that value with Nan and droping that
df.replace("?", np.nan, inplace=True)
df.dropna(inplace=True)
```

```python
# Count zeros in each column
zero_counts = (df == 0).sum()

print(zero_counts)
```

```python
zero_percentage = (df == 0).mean() * 100

zero_percentage = zero_percentage[zero_percentage > 0]

zero_percentage.plot(kind='bar', figsize=(10, 5))

plt.title('Percentage of Zeros per Column')
plt.xlabel('Column')
plt.ylabel('Zeros (%)')
plt.xticks(rotation=45)
plt.show()
```
