```python
df = pd.read_csv("../../Datasets/house_price_dataset_original_v2_with_categorical_features.csv")
```

```python
#Checking if we have empty rows
df.isna().sum()
```

```python
# do we have duplicates?
int(df.duplicated().sum())
```

```python
#Let's remove the unncessary words from the columns
#And convert it to int and float

df['land_size_sqm'] = df['land_size_sqm'].str.replace(' sqm', '', regex=False).astype(int)
df['house_size_sqm'] = df['house_size_sqm'].str.replace(' sqm', '', regex=False).astype(int)
df['distance_to_school'] = df['distance_to_school'].str.replace('km', '', regex=False).astype(float)
df['house_age'] = df['house_age'].str.replace('Years', '', regex=False).astype(int)
df['distance_to_supermarket'] = df['distance_to_supermarket'].str.replace('km', '', regex=False).astype(float)

```

```python
df.columns
```

```python
#Convert the binary columns of true/false to 1/0

varlist =  ['large_living_room', 'parking_space', 'front_garden', 'swimming_pool', 'wall_fence', 'water_front']

# Defining the map function
def binary_map(x):
    return x.map({'Yes': 1, "No": 0})

# Applying the function to the housing list
df[varlist] = df[varlist].apply(binary_map)
```

```python
#Checking how many different values does this column have
df['room_size'].unique()
```

```python
#Replacing the current values with the numeric value
category_mapper = {'extra_large': 3, 'large': 2, 'medium': 1, 'small': 0}
df['room_size'] = df['room_size'].replace(category_mapper)

#Converting to int
df['room_size'] = df['room_size'].astype(int)
```
