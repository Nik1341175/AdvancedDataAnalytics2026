### Ydata-profiling, regression example 1, housing data

```python
# pip install ydata-profiling

# if in Windows after installing ydata-profiling, you get: 
# ModuleNotFoundError: No module named 'pkg_resources'
# ---> install also the setuptools:
# python -m pip install "setuptools<82"

from ydata_profiling import ProfileReport
import pandas as pd

# open the data as a pandas DataFrame
df = pd.read_csv("../../Datasets/house_price_dataset_original_v2_with_categorical_features.csv")

# Generate the data profiling report, title is just for viewing purposes
report = ProfileReport(df,
    title='Seattle Housing Data, Data Engineering / Advanced Data Analytics, 2026'
)

# create the report into a separate HTML -file
# you could send this HTML-file to your colleaque 
# and it should work!
report.to_file("ydata_housing_regression_report.html")
```

```python
report = ProfileReport(df, title='Housing Data')
report.to_file("report.html")  # keep this for the record

# Get alerts directly as objects
alerts = report.get_description().alerts
for alert in alerts:
    print(alert)
```
